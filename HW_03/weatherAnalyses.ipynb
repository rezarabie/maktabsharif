```python
import pandas as pd
import numpy as np

cities=["Tehran"," Mashhad", "Isfahan", "Tabriz","Shiraz","Ahvaz"]
df=pd.DataFrame()
weather_lst=[]
weather={}
for i in range(30):
    for city in cities:
        temperature=np.random.randint(15,45)
        humidity=np.random.randint(20,80)
        rainfall=np.random.randint(0,50)
        weather={
            "city": city,
            "day": i+1,
            "temperature":temperature,
            "humidity":humidity,
            "rainfall":rainfall
        }
        weather_lst.append(weather)
        


dfs=pd.DataFrame(weather_lst)

print((dfs.groupby("city").value_counts()).to_string())


```
