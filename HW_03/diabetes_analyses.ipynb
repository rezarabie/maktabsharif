```python
import pandas as pd
import numpy as np

data=pd.read_csv("diabetes.csv")

dfs=pd.DataFrame(data)
# print (dfs.to_string())

# dfs.fillna(0,inplace=True)

col_1=dfs["Pregnancies"].mean()

col_len=len(dfs.columns)


for i in range(col_len):
    col_name=dfs.columns[i]
    avg=round(dfs[col_name].mean(),2)
    dfs[col_name].fillna(avg,inplace=True)
        


#print (dfs.columns[0])
print (dfs.head(5))

new_dfs=dfs[["Outcome","Insulin","Age","BMI","BloodPressure","Glucose"]]


print (new_dfs.corr())

```
